```python
import os
import json
import pandas as pd


jsons = os.listdir("../baselines/deepreview/DeepReviewer_14B.json")
csv_template = pd.read_csv("../results/final_deepreview_cal_flash/scores.csv")
```

```python
len(jsons)
```

```python
csv_template["pred_score"] = -100.0
```

```python
csv_template = csv_template.drop_duplicates(subset=["paper_id"], keep="first")
```

```python
len(csv_template)
```

```python
os.makedirs("../baselines_res/deepreview/reviews", exist_ok=True)
for json_file in jsons:
    with open(f"../baselines/deepreview/DeepReviewer_14B.json/{json_file}", "r") as f:
        data = json.load(f)
    try:
        content = data["results"][0]["meta_review"]["content"]
        with open(f"../baselines_res/deepreview/reviews/{json_file.replace('.json', '')}", "w") as f:
            f.write(content)
        csv_template.loc[csv_template["paper_id"] == json_file.replace('.txt.json', ''), "pred_score"] = float(data["results"][0]["meta_review"]["rating"])
    except KeyError:
        print(f"KeyError in {json_file}")
```

```python
csv_template.to_csv("../baselines_res/deepreview/scores.csv")
```

```python
import os
import pandas as pd
import re

openreview = os.listdir("../baselines_res/openreviewer")
csv_template = pd.read_csv("../results/final_deepreview_cal_flash/scores.csv")
csv_template["pred_score"] = -100.0

os.makedirs("../results/openreviewer/reviews", exist_ok=True)
for review in openreview:
    with open(f"../baselines_res/openreviewer/{review}", "r") as f:
        content = f.read()
    with open(f"../results/openreviewer/reviews/{review}", "w") as f:
        f.write(content)
    ans = float(re.sub(r'\D', '', content.split("## Rating")[-1].split("## Confidence")[0].strip()))
    if ans > 10:
        ans = -100
    csv_template.loc[csv_template["paper_id"] == review.replace('.txt', ''), "pred_score"] = ans
```

```python
csv_template.to_csv("../results/openreviewer/scores.csv") 
```
